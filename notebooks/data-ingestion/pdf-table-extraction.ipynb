# Extracting tables from PDF files
**Problem:** Tabular data is often locked inside PDF reports. This notebook pulls it out into a DataFrame or CSV.
**Approach:** `tabula.read_pdf(..., pages='all')` returns one DataFrame per detected table. The first one is saved to CSV.
**Requirements:** a Java runtime (`sudo apt install default-jre`) and `pip install tabula-py`.
**Data:** `datasets/sample-pdfs/sample_table.pdf`, a synthetic report generated by `make_sample_table_pdf.py`. Set `pdf_path` to try your own file.

```python
#require sudo apt install default-jre
#require pip install tabula-py
import tabula
```

```python
# Synthetic sample report (datasets/sample-pdfs/); point this at your own PDF
pdf_path = "../../datasets/sample-pdfs/sample_table.pdf"
dfs = tabula.read_pdf(pdf_path, pages='all')
print(f"{len(dfs)} table(s) found")
```

```python
# Save the first table as CSV
dfs[0].to_csv("sample_table.csv", index=False)
```
