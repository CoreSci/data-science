# Crypto lending yield analysis
**Problem:** What did lending stablecoins on an exchange actually return over a period, net of fees?
**Approach:** Load the lending history, filter the USDT loans, sum the net `Earned` interest, express it as a % of the principal, and measure the period covered. The deposit history is loaded alongside for reference.
**Data:** **Synthetic** sample files in `datasets/synthetic/crypto-lending/`, generated to mirror the columns of a Poloniex lending/deposit export. They contain no real balances, transactions or wallet addresses. Replace them with your own export to analyse real activity.

```python
import pandas as pd
```

```python
# Synthetic sample data (same columns as an exchange lending/deposit export, fake values)
df = pd.read_csv('../../datasets/synthetic/crypto-lending/lending.csv', parse_dates=['Close'], index_col='Close')
df1 = pd.read_csv('../../datasets/synthetic/crypto-lending/deposits.csv', index_col='Date')
```

```python
df
```

```python
USD = df["Currency"] == "USDT"
```

```python
dfUSD = df[USD]
```

```python
earningUSD = dfUSD["Earned"].sum()
```

```python
USDT_PRINCIPAL = 5000.00  # synthetic: USDT capital made available for lending
earningUSD / USDT_PRINCIPAL *100  # return on principal over the period, in %
```
