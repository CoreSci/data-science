# ETF relative-strength ratios
**Problem:** Compare Canadian ETFs head-to-head. A rising A/B price ratio means A is outperforming B.
**Approach:** Download 3 months of prices for five ETFs, plot the raw closes, then plot pairwise close-price ratios.
**Data:** Public Yahoo Finance data for VGRO.TO, VDY.TO, ZRR.TO, ZHP.TO and ZBAL.TO.

```python
%matplotlib inline
import matplotlib.pyplot as plt
from matplotlib import rc
from pylab import rcParams
import seaborn as sns

import yfinance as yf

sns.set(style='whitegrid', palette='muted', font_scale=1.5)
rcParams['figure.figsize'] = 14, 8
```

```python
ticker1 = "VGRO.TO"
stock1 = yf.Ticker(ticker1)
stockHistory1 = stock1.history(period="3mo")

ticker2 = "VDY.TO"
stock2 = yf.Ticker(ticker2)
stockHistory2 = stock2.history(period="3mo")

ticker3 = "ZRR.TO"
stock3 = yf.Ticker(ticker3)
stockHistory3 = stock3.history(period="3mo")

ticker4 = "ZHP.TO"
stock4 = yf.Ticker(ticker4)
stockHistory4 = stock4.history(period="3mo")

ticker5 = "ZBAL.TO"
stock5 = yf.Ticker(ticker5)
stockHistory5 = stock5.history(period="3mo")
```

```python
ratio1 = stockHistory1["Close"]/stockHistory3["Close"]

ratio2 = stockHistory2["Close"]/stockHistory3["Close"]

ratio3 = stockHistory4["Close"]/stockHistory1["Close"]

ratio4 = stockHistory4["Close"]/stockHistory2["Close"]

ratio5 = stockHistory3["Close"]/stockHistory4["Close"]
```

```python
plt.plot(stockHistory1.index, stockHistory1["Close"], label=ticker1, color="blue")
plt.plot(stockHistory2.index, stockHistory2["Close"], label=ticker2, color='green')
plt.plot(stockHistory3.index, stockHistory3["Close"], label=ticker3, color='red')
plt.plot(stockHistory4.index, stockHistory4["Close"], label=ticker4, color='purple')
plt.plot(stockHistory5.index, stockHistory5["Close"], label=ticker5, color='yellow')

plt.title('Canadian ETFs: daily close (last 3 months)')
plt.xlabel('Date')
plt.ylabel('Close')
plt.legend(loc='best')

plt.show();
```

```python
plt.plot(stockHistory3.index, ratio1, label=ticker1+" / "+ticker3, color='blue')
plt.plot(stockHistory3.index, ratio2, label=ticker2+" / "+ticker3, color='green')

plt.title('Stocks Comparative Ratio')
plt.xlabel('Date')
plt.ylabel('Stock A Close / Stock B Close')
plt.legend(loc='best')

plt.show();
```

```python
plt.plot(stockHistory4.index, ratio3, label=ticker4+" / "+ticker1, color='red')
plt.plot(stockHistory4.index, ratio4, label=ticker4+" / "+ticker2, color='purple')

plt.title('Stocks Comparative Ratio')
plt.xlabel('Date')
plt.ylabel('Stock A Close / Stock B Close')
plt.legend(loc='best')

plt.show();
```

```python
plt.plot(stockHistory3.index, ratio5, label=ticker3+" / "+ticker4, color='gold')

plt.title('Stocks Comparative Ratio')
plt.xlabel('Date')
plt.ylabel('Stock A Close / Stock B Close')
plt.legend(loc='best')

plt.show();
```
