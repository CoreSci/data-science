# Cross-asset correlation matrix
**Problem:** How closely does a portfolio fund move with broad indices, commodities, volatility, rates, currency and crypto?
**Approach:** Download 6 months of closing prices for 12 tickers, align them into one DataFrame, compute the Pearson correlation matrix and save it to a dated CSV.
**Data:** Public Yahoo Finance data (VGRO.TO, Wilshire 5000, S&P 500, Dow, Nasdaq futures, crude oil, VIX, gold, 10-year yield, CAD/USD, S&P/TSX, BTC).

```python
import pandas as pd
import yfinance as yf
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ticker ="VGRO.TO"
period = "6mo"
```

```python
ticker1 = ticker
ticker2 = "^W5000"
ticker3 = "^GSPC"
ticker4 = "^DJI"
ticker5 = "NQ=F"
ticker6 = "CL=F"
ticker7 = "^VIX"
ticker8 = "GC=F"
ticker9 = "^TNX"
ticker10 = "CADUSD=X"
ticker11 = "^GSPTSE"
ticker12 = "BTC-USD"
```

```python
stock1 = yf.Ticker(ticker1)
stock2 = yf.Ticker(ticker2)
stock3 = yf.Ticker(ticker3)
stock4 = yf.Ticker(ticker4)
stock5 = yf.Ticker(ticker5)
stock6 = yf.Ticker(ticker6)
stock7 = yf.Ticker(ticker7)
stock8 = yf.Ticker(ticker8)
stock9 = yf.Ticker(ticker9)
stock10 = yf.Ticker(ticker10)
stock11 = yf.Ticker(ticker11)
stock12 = yf.Ticker(ticker12)
```

```python
stockHistory1 = stock1.history(period=period)
stockHistory2 = stock2.history(period=period)
stockHistory3 = stock3.history(period=period)
stockHistory4 = stock4.history(period=period)
stockHistory5 = stock5.history(period=period)
stockHistory6 = stock6.history(period=period)
stockHistory7 = stock7.history(period=period)
stockHistory8 = stock8.history(period=period)
stockHistory9 = stock9.history(period=period)
stockHistory10 = stock10.history(period=period)
stockHistory11 = stock11.history(period=period)
stockHistory12 = stock12.history(period=period)
```

```python
matrix = pd.concat([stockHistory1["Close"], stockHistory2["Close"], stockHistory3["Close"],
                    stockHistory4["Close"], stockHistory5["Close"], stockHistory6["Close"],
                    stockHistory7["Close"], stockHistory8["Close"],
                    stockHistory9["Close"], stockHistory10["Close"], stockHistory11["Close"],
                    stockHistory12["Close"]], axis=1)
```

```python
matrix.columns = [ticker1, ticker2, ticker3, ticker4, ticker5, ticker6, ticker7, ticker8, ticker9, ticker10, ticker11, ticker12]
```

```python
dateIndex = str(stockHistory1.index[-1])
dateIndex = dateIndex.split()[0]
```

```python
matrixCompare = matrix.corr()
df = pd.DataFrame(matrixCompare)
title = ticker+'_correlation_matrix_period='+period+'_'+dateIndex+'.csv'
df.to_csv(title)
```
