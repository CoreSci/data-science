# US macro indicators from FRED
**Problem:** See the macro backdrop (growth, inflation, policy rate, unemployment, industrial output, oil) on one set of charts before making portfolio decisions.
**Approach:** Pull seven FRED series through Nasdaq Data Link, rescale them onto a shared axis, and plot growth-oriented and inflation/rates-oriented indicators on two charts.
**Data:** Public FRED series (GDP, real GDP, CPI, effective Fed funds rate, unemployment rate, industrial production, WTI crude). The saved charts show that public data.
**Credentials:** supplied by you through environment variables. None are stored in this repo. Copy `.env.example`, fill it in, and export the variables before starting Jupyter:
```bash
# macOS / Linux
export NASDAQ_DATA_LINK_API_KEY="..."
# Windows PowerShell
$env:NASDAQ_DATA_LINK_API_KEY = "..."
```

````python
%matplotlib inline
import matplotlib.pyplot as plt
from pylab import rcParams
import seaborn as sns

import nasdaqdatalink
import pandas as pd

sns.set(style='whitegrid', palette='muted', font_scale=1.5)
rcParams['figure.figsize'] = 14, 8
````

````python
# API key comes from an environment variable (see .env.example):
#   export NASDAQ_DATA_LINK_API_KEY=...
import os
API_KEY = os.environ["NASDAQ_DATA_LINK_API_KEY"]

mydata0 = nasdaqdatalink.get("FRED/GDP", authtoken=API_KEY)
mydata1 = nasdaqdatalink.get("FRED/GDPC1", authtoken=API_KEY)
mydata2 = nasdaqdatalink.get("FRED/CPIAUCSL", authtoken=API_KEY)
mydata3 = nasdaqdatalink.get("FRED/DFF", authtoken=API_KEY)
mydata4 = nasdaqdatalink.get("FRED/UNRATE", authtoken=API_KEY)
mydata5 = nasdaqdatalink.get("FRED/INDPRO", authtoken=API_KEY)
mydata6 = nasdaqdatalink.get("FRED/DCOILWTICO", authtoken=API_KEY)
````

````python
plt.plot(mydata0.index, mydata0["Value"], label="GDP", color='blue')
plt.plot(mydata1.index, mydata1["Value"], label="Real GDP", color='green')
plt.plot(mydata5.index, mydata5["Value"]*200, label="Industrial Index X200", color='red')
plt.plot(mydata6.index, mydata6["Value"]*200, label="Crude Oil WTI X200", color='grey')

plt.title('US growth indicators: GDP, real GDP, industrial production and WTI (rescaled)')
plt.xlabel('Date')
plt.ylabel('USD billions / rescaled index')
plt.legend(loc='best')

plt.show();
````

````python
plt.plot(mydata2.index, mydata2["Value"]/10, label="Consumer Price Index / 10", color='green')
plt.plot(mydata3.index, mydata3["Value"], label="Fed funds rate (%)", color='blue')
plt.plot(mydata4.index, mydata4["Value"], label="Unemployment rate (%)", color='red')

plt.title('US inflation, policy rate and unemployment')
plt.xlabel('Date')
plt.ylabel('percent / rescaled index')
plt.legend(loc='best')

plt.show();
````
