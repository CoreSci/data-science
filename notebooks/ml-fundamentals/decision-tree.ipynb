# Decision tree classifier: minimal example
**Problem:** Show the fit/predict workflow of a decision tree on the smallest possible dataset.
**Approach:** Train `DecisionTreeClassifier` on four labelled 2-feature samples and classify one new point.
**Data:** Four inline samples (sepal length/width-style features).

Simple example of a decision tree using Python with the popular scikit-learn library. This is a basic example using two features (e.g., sepal length and sepal width) to classify samples into two classes. You can adapt it to your specific dataset and problem.

```python
from sklearn import tree
```

```python
# Sample data
features = [[5.1, 3.5], [4.9, 3.0], [6.7, 3.1], [6.0, 3.0]]
labels = ['Class1', 'Class1', 'Class2', 'Class2']
```

```python
# Create and fit the decision tree model
clf = tree.DecisionTreeClassifier()
clf = clf.fit(features, labels)
```

```python
# Make predictions
new_data = [[5.5, 3.2]]
prediction = clf.predict(new_data)

print("Predicted class:", prediction)
```
