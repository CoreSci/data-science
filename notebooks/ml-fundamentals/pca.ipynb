# Principal component analysis (PCA)
**Problem:** Visualise 4-dimensional data in 2-D while keeping as much variance as possible.
**Approach:** Project the Iris features onto two principal components and scatter-plot them by class.
**Data:** scikit-learn's built-in Iris dataset.

Simple example of PCA (Principal Component Analysis) using scikit-learn in Python. In this example, we load the Iris dataset, apply PCA to reduce the dimensionality to 2 components, and then visualize the results with a scatter plot. The code uses scikit-learn's PCA module and matplotlib for plotting.

```python
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
# Load iris dataset
iris = load_iris()
X = iris.data
y = iris.target
```

```python
# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
```

```python
# Plot the results
plt.figure(figsize=(8, 6))
for i in range(len(np.unique(y))):
    plt.scatter(X_pca[y == i, 0], X_pca[y == i, 1], label=f'Class {i}')

plt.title('PCA of Iris Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```
