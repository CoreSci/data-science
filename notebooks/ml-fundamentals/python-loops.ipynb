# Python loop refresher
**Problem:** Minimal side-by-side reminder of `for` and `while` iteration.
**Approach:** Print the numbers 1-5 with both constructs.
**Data:** None.

```python
#for loop
numbers = [1, 2, 3, 4, 5]

for num in numbers:
    print(num)
```

```python
#while loop
counter = 1

while counter <= 5:
    print(counter)
    counter += 1
```
