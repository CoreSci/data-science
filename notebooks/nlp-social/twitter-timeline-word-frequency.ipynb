# Twitter timeline word frequency
**Problem:** What topics does an account keep tweeting about? Word frequencies over its recent timeline give a quick answer.
**Approach:** List the accounts you follow, pick one, pull its recent timeline with `python-twitter`, then tokenise with NLTK, keep alphabetic lower-cased tokens, remove English + custom stop words (`stop_words_addon1.py`) and rank the rest with `FreqDist`. The last cell prints candidate stop words in a paste-ready format, which helps grow the custom list.
**Data:** Your own authenticated Twitter/X account. Outputs are cleared here, so run it with your own credentials.
**Note:** Written against the Twitter v1.1 API. Current X API access tiers may restrict these endpoints.
**Credentials:** supplied by you through environment variables. None are stored in this repo. Copy `.env.example`, fill it in, and export the variables before starting Jupyter:
```bash
# macOS / Linux
export TWITTER_CONSUMER_KEY="..."
# Windows PowerShell
$env:TWITTER_CONSUMER_KEY = "..."
```

````python
import os
import twitter
import pandas as pd

import nltk
from nltk.corpus import stopwords
from nltk import FreqDist

import stop_words_addon1 as swtb
````

````python
# Credentials come from environment variables (see .env.example); never hard-code them.
#   export TWITTER_CONSUMER_KEY=...  TWITTER_CONSUMER_SECRET=...
#   export TWITTER_ACCESS_TOKEN=...  TWITTER_ACCESS_TOKEN_SECRET=...
api = twitter.Api(consumer_key=os.environ["TWITTER_CONSUMER_KEY"],
                  consumer_secret=os.environ["TWITTER_CONSUMER_SECRET"],
                  access_token_key=os.environ["TWITTER_ACCESS_TOKEN"],
                  access_token_secret=os.environ["TWITTER_ACCESS_TOKEN_SECRET"])
````

````python
users = api.GetFriends()
````

````python
usersID = []
usersName = []
i = 0
while i < len(users):
    usersID.append(users[i].id)
    usersName.append(users[i].name)
    i = i + 1


df1 = pd.DataFrame(usersID, columns=["ID"])
df2 = pd.DataFrame(usersName, columns=["name"])

df = pd.concat([df1, df2], axis=1)
````

````python
i = 0
while i < len(df):
    print(i, ' : ', df['name'][i])
    i = i + 1
````

````python
nameID = 0  # index of the followed account to analyse (see the list printed above)
````

````python
df['name'][nameID]
````

````python
twitTime = api.GetUserTimeline(df["ID"][nameID])
````

````python
sentence = ""
````

````python
i = 0

while i < len(twitTime):
    sentence = sentence + " " + twitTime[i].text
    i = i + 1
````

````python
tokens0 = nltk.word_tokenize(sentence)
````

````python
tokens1 = [word for word in tokens0 if word.isalpha()]
````

````python
tokens2 = [w.lower() for w in tokens1]
````

````python
stop_words = stopwords.words('english')
````

````python
stop_words = swtb.add_stop_words(stop_words)
````

````python
tokens3 = [w for w in tokens2 if not w in stop_words]
````

````python
print(len(tokens0))
print(len(tokens1))
print(len(tokens2))
print(len(tokens3))
````

````python
freq_dist_pos = FreqDist(tokens3)
i = 0
for items in freq_dist_pos.most_common(1000):
    print(i, " : ", items)
    i=i+1
````

````python
# paste-ready candidates for extending stop_words_addon1.py
for element in freq_dist_pos:
    print("stop_words.append('"+element+"')")
````
