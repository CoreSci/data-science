# OpenCV image loading and display
**Problem:** Load an image with OpenCV and display it inside a notebook, including in a headless Docker Jupyter container where `cv2.imshow` isn't available.
**Approach:** `cv2.imread` → convert BGR to RGB → display with matplotlib.
**Data:** A synthetic sample image in `datasets/sample-images/`. Point `image_path` at any image to try your own.

Load the image with OpenCV. `cv2.imread` returns a NumPy array in **BGR** channel order.

```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Load an image from file
image_path = '../../datasets/sample-images/sample_image.jpg'  # synthetic sample image; replace with your own
img = cv2.imread(image_path)
```

`cv2.imshow` opens a native window, which isn't available in a headless Docker/Jupyter session. So the image is converted to RGB and displayed with matplotlib instead.

```python
#Use matplotlib here.
# Convert BGR image to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# Display the image using Matplotlib
plt.imshow(img_rgb)
plt.title('Sample image (converted BGR -> RGB)')
plt.axis('off')  # hide pixel axes
plt.show()
```
