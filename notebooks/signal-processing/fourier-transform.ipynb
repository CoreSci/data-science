# Fourier transform: time domain → frequency domain
**Problem:** Identify which frequencies make up a signal.
**Approach:** A reusable `apply_fourier_transform()` helper computes the FFT with NumPy, normalises the magnitude and plots the time-domain signal next to its positive-frequency spectrum.
**Data:** A synthetic 1-second signal sampled at 1 kHz, mixing 50 Hz and 120 Hz sine waves. The two peaks in the spectrum confirm the method.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def apply_fourier_transform(signal, sampling_rate):
    """
    Applies Fourier Transform to a given signal and plots the results.
    
    Parameters:
    signal (numpy array): Time-domain signal
    sampling_rate (float): Sampling frequency in Hz
    """
    # Compute FFT
    N = len(signal)
    freq_domain = np.fft.fft(signal)
    freq_magnitude = np.abs(freq_domain) / N  # Normalize amplitude
    freq = np.fft.fftfreq(N, d=1/sampling_rate)  # Frequency axis
    
    # Plot time-domain signal
    time = np.arange(N) / sampling_rate
    plt.figure(figsize=(12, 6))
    plt.subplot(2, 1, 1)
    plt.plot(time, signal)
    plt.title("Time-Domain Signal")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    
    # Plot frequency-domain representation
    plt.subplot(2, 1, 2)
    plt.plot(freq[:N//2], freq_magnitude[:N//2])  # Only positive frequencies
    plt.title("Frequency-Domain Representation")
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Magnitude")
    
    plt.tight_layout()
    plt.show()
```

```python
# Example usage
if __name__ == "__main__":
    sampling_rate = 1000  # Hz
    t = np.linspace(0, 1, sampling_rate, endpoint=False)  # 1 second duration
    signal = np.sin(2 * np.pi * 50 * t) + 0.5 * np.sin(2 * np.pi * 120 * t)  # Example signal with 50Hz & 120Hz components
    apply_fourier_transform(signal, sampling_rate)
```
