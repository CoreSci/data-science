# Preparing time-series windows for an LSTM
**Problem:** Recurrent models need the input as `(samples, time_steps, features)`. This notebook converts a 1-D series into supervised windows.
**Approach:** Generate a sine wave, slide a 10-step window over it (each window predicts the next value), then reshape for Keras.
**Data:** 100 synthetic points of sin(x).
**Status:** Data preparation only. The model definition and training steps are not implemented yet.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
# Generate some sample data (sin wave)
def generate_data(n_points):
    x = np.linspace(0, 10, n_points)
    y = np.sin(x)
    return y
```

```python
# Prepare data for time series analysis
def prepare_data(data, n_steps):
    X, y = [], []
    for i in range(len(data)):
        end_ix = i + n_steps
        if end_ix > len(data)-1:
            break
        seq_x, seq_y = data[i:end_ix], data[end_ix]
        X.append(seq_x)
        y.append(seq_y)
    return np.array(X), np.array(y)
```

```python
# Define parameters
n_steps = 10
n_features = 1  # univariate time series (1 feature)
n_units = 50    # number of units in LSTM layer
```

```python
# Generate and prepare data
data = generate_data(100)
X, y = prepare_data(data, n_steps)
X = X.reshape((X.shape[0], X.shape[1], n_features))
```
